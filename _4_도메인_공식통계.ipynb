```python
!pip install pyreadstat

import pandas as pd
import pyreadstat

# 1. 구글 드라이브 연동 (이미 연동되어 있다면 생략 가능)
from google.colab import drive
drive.mount('/content/drive')

# 2. 원자료 불러오기 (SAS 데이터 파일 경로 설정)
file_path = '/content/drive/MyDrive/hdac_data/kyrbs2025.sas7bdat'
cols = ['GRADE', 'F_BR', 'W', 'HT', 'WT']
df, meta = pyreadstat.read_sas7bdat(file_path, usecols=cols)

# 3. 고등학생 도메인 추출 (GRADE 4: 고1, 5: 고2, 6: 고3)
df_high = df[df['GRADE'].isin([4, 5, 6])].copy()

# 4. 아침식사 결식 변수(F_BR) 재코딩
# 이용지침서 기준 주 5회 이상 결식(원 코드 1, 2, 3)은 1, 나머지는 0으로 설정
df_high['breakfast_skipping'] = df_high['F_BR'].apply(lambda x: 1 if x in [1, 2, 3] else (0 if x >= 4 else float('nan')))

# 5. 복합표본 가중치(W)를 반영한 가중 결식률 자동 계산
valid_df = df_high.dropna(subset=['breakfast_skipping', 'W'])
weighted_skip_sum = valid_df[valid_df['breakfast_skipping'] == 1]['W'].sum()
weighted_total_sum = valid_df['W'].sum()

weighted_skipping_rate = (weighted_skip_sum / weighted_total_sum) * 100

# 6. 결과 출력
print("="*40)
print(f"고등학생 도메인 가중 아침 결식률: {weighted_skipping_rate:.2f}%")
print("="*40)
```
